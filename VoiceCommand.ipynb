```python
# Instalar as bibliotecas necessárias
!pip install gTTS googletrans==4.0.0-rc1

from gtts import gTTS
from googletrans import Translator
from IPython.display import Audio

# Texto original em inglês
text_to_say = "Open youtube, play rock!"

# Traduzir o texto para português
translator = Translator()
translated_text = translator.translate(text_to_say, src='en', dest='pt').text

print("Texto traduzido:", translated_text)

# Configurar o idioma para português
language = "pt"

# Criar o objeto gTTS com o texto traduzido
gtts_object = gTTS(text=translated_text,
                   lang=language,
                   slow=False)

# Salvar o arquivo de áudio
gtts_object.save("/content/gtts.wav")

# Reproduzir o áudio
Audio("/content/gtts.wav")
```

```python
# Função para capturar áudio e converter em texto
def get_audio(file_path):
    r = sr.Recognizer()
    with sr.AudioFile(file_path) as source:
        audio = r.record(source)
        try:
            return r.recognize_google(audio, language="pt-BR").lower()
        except sr.UnknownValueError:
            print("Desculpe, não consegui entender o áudio.")
        except sr.RequestError:
            print("Erro ao conectar ao serviço de reconhecimento de voz.")
    return ""

# Função para falar (usando gTTS)
def speak(text):
    tts = gTTS(text=text, lang='pt')
    filename = "/content/voice.mp3"
    tts.save(filename)
    display(Audio(filename, autoplay=True))

# Função para gerar link do YouTube
def search_youtube(audio_text):
    words = audio_text.split()
    if "youtube" in words:
        index = words.index("youtube") + 1
        if index < len(words):
            keyword = " ".join(words[index:])
            url = f"https://www.youtube.com/results?search_query={keyword}"
            print(f"Clique aqui para abrir no YouTube: {url}")
            speak(f"Aqui está o link para {keyword} no YouTube.")
        else:
            speak("Nenhuma palavra-chave foi encontrada para pesquisa.")

# Processa o áudio já gravado
audio_text = get_audio("/content/audio.wav")
if audio_text:
    print("Texto reconhecido:", audio_text)
    search_youtube(audio_text)
```
